# Kenya Migration Policy Analysis (2021–2024)
This notebook performs EDA and regression using UNHCR and OpenBudget Kenya data.

```python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
data = pd.read_csv('../data/kenya_real_2021_2024.csv')
data.head()
```

```python
plt.plot(data['Year'], data['Policy_Funding_BnKES'], marker='o', label='Funding')
plt.plot(data['Year'], data['Refugee_Population']/100000, marker='o', label='Refugee Pop (x100k)')
plt.legend()
plt.title('Funding vs Refugee Population (2021–2024)')
plt.show()
```

```python
X = sm.add_constant(data['Policy_Funding_BnKES'])
y = data['Refugee_Population']
model = sm.OLS(y, X).fit(cov_type='HC3')
print(model.summary())
```
